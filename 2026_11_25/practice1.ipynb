```python
def split_bill(total,people):
    try:
        per_person = total / people
        print(f"每人應付{per_person:.0f}元")
    except ZeroDivisionError:
        print("錯誤:分攤人數不能為0")
    except :
        print("其他錯誤")
        

split_bill(1000,3)
```

```python
def calc_unit_print(total_text,count_text):
    try:
        total=int(total_text)
        count=int(count_text)
        unit_price =  total / count
        print(f"單價為:{unit_price:.1f}元")
    except ValueError as e:
        print(f"輸入格式錯誤，請輸入整數!(系統訊息{e})")
    except ZeroDivisionError:
        print("數量不能為0，無法計算")
    
calc_unit_print("500","4")
calc_unit_print("5百","4")
```

```python
def get_item(data,index_text):
    try:
        index=int(index_text)
        print("取得項目:",data[index])
    except (ValueError,IndexError) as e:
        print(f"查詢失敗!(輸入非整數或索引超出範圍):{e}")
    except Exception as e:
        print(f"發生未預期的錯誤[{type(e)._name_}]:(e)")

items=['蘋果','香蕉','橘子']
get_item(items,1)
get_item(items,12345)
get_item(items,'青蘋果')
```
